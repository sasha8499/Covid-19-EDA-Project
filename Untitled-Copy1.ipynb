```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")
```

# DATA CLEANING

```python
df = pd.read_csv("worldometer_coronavirus_daily_data.csv")
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df['daily_new_cases'].fillna(0,inplace=True)
```

```python
df['daily_new_deaths'].fillna(0,inplace=True)
```

```python
df['active_cases'].fillna(method='ffill',inplace=True)
```

```python
df['cumulative_total_deaths'].fillna(method='ffill',inplace=True)
```

```python
df.isnull().sum()
```

```python
duplicates = df.duplicated().sum()
duplicates
```

```python
for col in df.select_dtypes(include="object").columns:
    print(f"{col}: {df[col].nunique()} unique values")
```

```python
numeric_cols = df.select_dtypes(include='number').columns

for col in numeric_cols:
    negatives = (df[col] < 0).sum()
    if negatives > 0:
        print(f"{col}: {negatives} negative values")
```

```python
df['daily_new_cases'] = df['daily_new_cases'].apply(lambda x: max(x, 0))
df['daily_new_deaths'] = df['daily_new_deaths'].apply(lambda x: max(x, 0))
df['active_cases'] = df['active_cases'].apply(lambda x: max(x, 0))
```

```python
df.describe()
```

```python
print("Unique countries:", df['country'].nunique())
print("Date range:", df['date'].min(), "to", df['date'].max())
print(df.dtypes)
```

```python
# Save cleaned DataFrame to CSV
df.to_csv('cleaned_covid19_data.csv', index=False)
```

# EXPLORATORY DATA ANALYSIS(EDA)

```python
global_cases = df.groupby('date')['daily_new_cases'].sum()

plt.figure(figsize=(14,6))
plt.plot(global_cases, color='steelblue')
plt.title('Global Daily New COVID-19 Cases Over Time')
plt.xlabel('Date')
plt.ylabel('Cases')
plt.tight_layout()
plt.show()
```

```python
global_deaths = df.groupby('date')['daily_new_deaths'].sum()

plt.figure(figsize=(14,6))
plt.plot(global_deaths, color='darkred')
plt.title('Global Daily New COVID-19 Deaths Over Time')
plt.xlabel('Date')
plt.ylabel('Deaths')
plt.tight_layout()
plt.show()
```

```python
latest_data = df[df['date'] == df['date'].max()]

top_countries_cases = latest_data.groupby('country')['cumulative_total_cases'].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(12,6))
sns.barplot(x=top_countries_cases.values, y=top_countries_cases.index, palette='coolwarm')
plt.title('Top 10 Countries by Total COVID-19 Cases')
plt.xlabel('Total Cases')
plt.ylabel('Country')
plt.tight_layout()
plt.show()
```

```python
india_data = df[df['country'] == 'India']

plt.figure(figsize=(14,6))
plt.plot(india_data['date'], india_data['daily_new_cases'], label='Daily New Cases', color='blue')
plt.plot(india_data['date'], india_data['daily_new_deaths'], label='Daily New Deaths', color='red')
plt.title('COVID-19 Trend in India')
plt.xlabel('Date')
plt.ylabel('Count')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Convert to datetime
df['date'] = pd.to_datetime(df['date'])

# Extract month period
df['month'] = df['date'].dt.to_period('M')

# Group and visualize
monthly_trend = df.groupby('month')[['daily_new_cases', 'daily_new_deaths']].sum()

monthly_trend.plot(kind='bar', figsize=(14,6), color=['skyblue', 'salmon'])
plt.title('Monthly COVID-19 Cases & Deaths (Worldwide)')
plt.xlabel('Month')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(df.select_dtypes(include='number').corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.tight_layout()
plt.show()
```

```python
# Group by country and sum the total deaths
top_deaths = df.groupby('country')['cumulative_total_deaths'].max().sort_values(ascending=False).head(10)

# Plot
plt.figure(figsize=(12,6))
top_deaths.plot(kind='bar', color='crimson')
plt.title('Top 10 Countries by Total COVID-19 Deaths')
plt.xlabel('Country')
plt.ylabel('Total Deaths')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Step 1: Top 5 countries by total cases
top5_countries = df.groupby('country')['cumulative_total_cases'].max().sort_values(ascending=False).head(5).index.tolist()

# Step 2: Filter dataset
top5_data = df[df['country'].isin(top5_countries)].copy()
top5_data['date'] = pd.to_datetime(top5_data['date'])

# Step 3: Group by month and country
top5_data['month'] = top5_data['date'].dt.to_period('M')
monthly_top5 = top5_data.groupby(['month', 'country'])['daily_new_cases'].sum().reset_index()

# Step 4: Pivot for plotting
monthly_pivot = monthly_top5.pivot(index='month', columns='country', values='daily_new_cases')

# Step 5: Plot
monthly_pivot.plot(figsize=(14,6), marker='o')
plt.title('Monthly COVID-19 Cases – Top 5 Countries')
plt.xlabel('Month')
plt.ylabel('Monthly Cases')
plt.xticks(rotation=45)
plt.legend(title='Country')
plt.tight_layout()
plt.show()
```
